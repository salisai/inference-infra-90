```python
import torch

p = torch.cuda.get_device_properties(0)
print("name:", p.name)
print("SMs:", p.multi_processor_count)
print("memory GB:", round(p.total_memory / 1e9, 1))
print("compute capability:", p.major, p.minor)
```

```python
def occupancy(threads_per_block, regs_per_thread,
              max_threads=2048, max_blocks=32, regs_per_sm=65536):
    by_threads = max_threads // threads_per_block
    by_regs = regs_per_sm // (threads_per_block * regs_per_thread)
    blocks = min(by_threads, max_blocks, by_regs)
    return blocks, blocks * threads_per_block / max_threads

for regs in (32, 33, 64):
    print(regs, occupancy(512, regs))
```
